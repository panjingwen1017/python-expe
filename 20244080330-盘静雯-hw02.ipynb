```python
# 字符串的乐趣
def is_palindrome(s):
    clean = s.replace(" ", "").lower()
    return clean == clean[::-1]


def is_abecedarian(s):
    clean = s.replace(" ", "")
    for i in range(len(clean) - 1):
        if clean[i] > clean[i+1]:
            return False
    return True


def remove_vowels(s):
    vowels = {'a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U'}
    result = []
    for char in s:
        if char not in vowels:
            result.append(char)
    return "".join(result)

# 简单自测
print(is_palindrome("kayak"))
print(is_abecedarian("adder"))
print(remove_vowels("audio"))
```

```python
# 列表的乐趣
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("参数必须为列表")
    return lst[::-1]


def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i+1]:
            return False
    return True

# 简单自测
print(list_reverse([1,2,3]))
print(is_sorted([1,2,4,5]))
```

```python
# 作为向量的元组
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个输入必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量必须是整数或者浮点数")
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("元组内元素必须为数字")
    return tuple(x * s for x in t)


def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("输入必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量长度不一致")
    return float(sum(a * b for a, b in zip(t1, t2)))


def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    row_length = None
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if row_length is None:
            row_length = len(row)
        else:
            if len(row) != row_length:
                return False
        for item in row:
            if not isinstance(item, (int, float)):
                return False
    return True


def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise ValueError("矩阵格式不合法")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    if len(vec) != len(mx[0]):
        raise ValueError("矩阵列数和向量维度不匹配")
    output = []
    for r in mx:
        total = 0
        for a, b in zip(r, vec):
            total += a * b
        output.append(total)
    return tuple(output)

# 简单自测
print(vec_scalar_mult((1,2,3), 2))
print(vec_inner_product((1,2),(3,4)))
print(check_valid_mx(((1,2),(3,4))))
```

```python
# 更多向量的乐趣
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能是负数")
        if elements is None:
            elements = [0] * dim
        if len(elements) != dim:
            raise ValueError("元素个数与维度不匹配")
        self.dim = dim
        self.elements = tuple(elements)

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("运算对象必须是Vector实例")
        if self.dim != other.dim:
            raise ValueError("向量维度不一致")
        return float(sum(x * y for x, y in zip(self.elements, other.elements)))

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements > other.elements

    def __ge__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements >= other.elements

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            new_elems = [x * other for x in self.elements]
            return Vector(self.dim, new_elems)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量维度不一致，无法相乘")
            new_elems = [a * b for a, b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            return NotImplemented

    def __rmul__(self, other):
        return self * other


# 简单自测
v1 = Vector(2, [2,4])
v2 = Vector(2, [2,5])
print(v1 < v2)
print(v1.dot(v2))
```
