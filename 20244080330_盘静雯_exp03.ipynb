```python
import re

# 任务1
s1 = "cat on mat; later: cats and dogs"
print("----------任务1----------")
print(r"\d+ 匹配数字：", re.findall(r"\d+", s1))
print(r"\s+ 匹配空白：", re.findall(r"\s+", s1))
print(r"\bcat\b 完整单词cat：", re.findall(r"\bcat\b", s1))

# 任务2
s2 = "cat dog bird CAT"
print("\n----------任务2----------")
print(r"cat|dog：", re.findall(r"cat|dog", s2))
print(r"a|aa|aaa 对'aaaa'：", re.findall(r"a|aa|aaa", "aaaa"))
print(r"aaa|aa|a 对'aaaa'：", re.findall(r"aaa|aa|a", "aaaa"))

# 任务3
s3 = "ct cat caat caaat"
print("\n----------任务3----------")
print("ca*t  ：", re.findall(r"ca*t", s3))
print("ca+t  ：", re.findall(r"ca+t", s3))
print("ca?t  ：", re.findall(r"ca?t", s3))
print("ca{2}t：", re.findall(r"ca{2}t", s3))
print("ca{1,2}t：", re.findall(r"ca{1,2}t", s3))
```

## 实验1：正则表达式的基础匹配
导入模块：`import re`

任务说明：
1. 对字符串`cat on mat; later: cats and dogs`分别测试元字符：
- `\d+`：匹配1个及以上数字
- `\s+`：匹配1个及以上空白字符（空格、制表符、换行）
- `\bcat\b`：`\b`单词边界，只匹配完整单词cat，不匹配cats

2. 分支选择`|`；对比两种贪婪匹配顺序：`a|aa|aaa`与`aaa|aa|a`，正则**从左向右匹配，匹配成功即返回**，因此顺序会影响结果。

3. 量词测试：
- `*`：前面字符出现0次或多次
- `+`：前面字符出现1次或多次
- `?`：前面字符出现0次或1次
- `{n}`：精确n次
- `{n,m}`：n～m次


```python
import re

# 任务1：提取邮箱用户名、域名
text_mail = "联系:alice@example.com;备选:bob@test.org"
print("----------任务1 邮箱捕获组----------")
mail_pat = r"(\w+)@(\w+\.\w+)"
mails = re.findall(mail_pat, text_mail)
for user, domain in mails:
    print(f"用户名:{user}, 域名:{domain}")

# 任务2：后向引用找连续重复单词
print("\n----------任务2 重复单词 后向引用----------")
text_word = "cat cat, dog cat, bird bird"
dup_pat = r"\b(\w+)\s+\1\b"
res_dup = re.findall(dup_pat, text_word)
print("连续重复的单词：", res_dup)

# 任务3：日期格式转换 年‑月‑日 → 月‑日‑年
print("\n----------任务3 日期转换----------")
text_date = "2024-05-01\n2024/05/02\n2024.05.03"
print("转换前：")
print(text_date)
date_pat = r"(\d{4})[-/.](\d{2})[-/.](\d{2})"
text_date_new = re.sub(date_pat, r"\2-\3-\1", text_date)
print("\n转换后：")
print(text_date_new)

# 任务4：电话号码统一格式 (123) 456‑7890
print("\n----------任务4 电话号码格式化----------")
text_tel = "联系我:123-456-7890 或 987.654.3210"
tel_pat = r"(\d{3})[-.](\d{3})[-.](\d{4})"
text_tel_new = re.sub(tel_pat, r"(\1) \2-\3", text_tel)
print("转换前：", text_tel)
print("转换后：", text_tel_new)
```

## 实验2：分组、回溯引用与替换
- `()`捕获组：括号把匹配内容分组，`re.findall`遇到捕获组会返回分组元组。
- 回溯引用 `\1`：引用第1个捕获组匹配到的内容，用来查找重复单词。
- `re.sub(pattern, repl, text)`：查找并替换文本，`\1 \2 \3`在替换字符串中代表各个捕获组。

任务：
1. 邮箱：`(\w+)@(\w+\.\w+)`，第1组用户名，第2组域名。
2. 重复单词：`\b(\w+)\s+\1\b`，捕获单词，后面再次引用该单词，匹配连续重复。
3. 日期转换：捕获年、月、日，把 `年‑月‑日` 重排为 `月‑日‑年`。
4. 电话号码格式化：捕获3段数字，统一输出`(xxx) xxx‑xxxx`格式。


```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body> <h1>数据分析</h1> <p class="course">第六讲:正则表达式</p> <p class="course">第七讲:结构化数据</p> <a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a> </body></html>
"""

# 任务1
soup = BeautifulSoup(html_doc, "html.parser")
print("----------任务1----------")
print("页面标题文本：", soup.title.get_text())
print("h1标签名：", soup.h1.name)
print("h1文本：", soup.h1.get_text())

# 任务2 find_all
print("\n----------任务2----------")
p_list = soup.find_all("p", class_="course")
print("class=course的段落：")
for p in p_list:
    print("  ", p.get_text())

a_list = soup.find_all("a")
print("\n所有链接：")
for a in a_list:
    print(f"  链接文字:{a.get_text()}, href:{a['href']}")

# 任务3 get_text对比
print("\n----------任务3 get_text对比----------")
print("整个文档get_text():")
print(repr(soup.get_text()))
print("\n单个p段落get_text():", repr(p_list[0].get_text()))
print("单个a链接get_text():", repr(a_list[0].get_text()))
```
