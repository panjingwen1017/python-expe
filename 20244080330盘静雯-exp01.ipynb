```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("分数必须在0~100之间")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

test = [-10,  40, 70, 85, 95, 105]
for s in test:
    try:
        print(s, score_level(s))
    except ValueError as err:
        print(s, err)
        
```

```python
def fact_iter(n):
    if n < 0:
        raise ValueError("非负整数")
    res = 1
    for k in range(2, n+1):
        res *= k
    return res

def fact_rec(n):
    if n < 0:
        raise ValueError("非负整数")
    if n == 0:
        return 1
    return n * fact_rec(n-1)

for x in [0,1,5,10]:
    print(f"n={x} 迭代:{fact_iter(x)} 递归:{fact_rec(x)}")
```

```python
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    mn = min(values)
    mx = max(values)
    avg = sum(values)/len(values)
    return mn, mx, avg

res = summary([85,58,92])
print(res)
mn, mx, avg = summary([85,58,92])
print(f"最小{mn},最大{mx},均值{avg:.2f}")
```

```python
import os
exp_dir = "exp01_data"
os.makedirs(exp_dir, exist_ok=True)
fp = os.path.join(exp_dir, "students.txt")
txt = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(fp, "w", encoding="utf-8") as f:
    f.write(txt)

# 读取
with open(fp, "r", encoding="utf-8") as f:
    lines = f.readlines()
for line in lines:
    print(line.strip())
```

```python
def parse_student(line):
    line = line.strip()
    if not line:
        return None
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError("字段数量不对")
    name, s_str = parts
    name = name.strip()
    if not name:
        raise ValueError("名字为空")
    s = int(s_str.strip())
    if not 0 <= s <=100:
        raise ValueError("分数超出范围0~100")
    return {"name": name, "score": s}

fp = os.path.join("exp01_data", "students.txt")
valid = []
invalid_log = []
with open(fp, "r", encoding="utf-8") as f:
    lines = f.readlines()[1:] #跳过表头

for idx, line in enumerate(lines, start=1):
    try:
        d = parse_student(line)
        if d:
            valid.append(d)
    except Exception as e:
        invalid_log.append( (idx, str(line).strip(), str(e)) )

print("有效记录：", valid)
print("无效记录：", invalid_log)

# 统计
if valid:
    s_list = [x["score"] for x in valid]
    avg = sum(s_list)/len(s_list)
    pass_num = sum(1 for x in s_list if x >= 60)
    pass_r = pass_num / len(s_list)
else:
    avg = 0
    pass_r = 0

report_path = os.path.join("exp01_data", "report.txt")
# w 覆盖
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效:{len(valid)},无效:{len(invalid_log)},平均分:{avg:.2f},及格率:{pass_r:.2%}\n")
# a追加
with open(report_path, "a", encoding="utf-8") as f:
    f.write("追加一行，a模式不会清空原有内容\n")

# pickle保存加载
pk_path = os.path.join("exp01_data", "students.pkl")
with open(pk_path, "wb") as f:
    pickle.dump(valid, f)
with open(pk_path, "rb") as f:
    load_back = pickle.load(f)
print("pickle恢复的数据和原数据相等：", load_back == valid)
```
