```python
# ========== 头部信息块 ==========
import os, uuid, hashlib, platform, getpass

# ===== 1. 填写学号 =====
STUDENT_ID = 20244080330  # 你的学号
# ===== 2. 打印 conda 环境 =====
CONDA_ENV = os.environ.get("CONDA_DEFAULT_ENV", "base")
print(f"[Conda 环境] {CONDA_ENV}")
# ===== 3. 机器指纹 =====
MACHINE_FP = hashlib.sha256(
    f"{platform.node()}|{platform.machine()}|{platform.system()}|{uuid.getnode()}".encode()
).hexdigest()[:16]
# ===== 4. 生成唯一 ID =====
RUN_ID = f"{STUDENT_ID}-{MACHINE_FP}"
print(f"[学号]    {STUDENT_ID}")
print(f"[机器指纹] {MACHINE_FP}")
print(f"[RUN_ID]  {RUN_ID}")
```

```python
# ========== 热身：简单函数 ==========
# 1 say_hi
def say_hi():
    print("Hello, world!")

# 2 goat_pad
def goat_pad(s):
    print("goat" + s + "goat")

# 3 print_n
def print_n(s, n):
    for _ in range(n):
        print(s)

# 测试调用
say_hi()
goat_pad("test")
print_n("hi", 3)
```

```python
# ========== 欧几里得算法 GCD ==========
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

# 题目要求的三组测试
print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
```

```python
# ========== 近似欧拉数 e ==========
import math

# 1 euler_limit: (1+1/n)^n
def euler_limit(n):
    return (1 + 1 / n) ** n

# 2 euler_infinite_sum
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1
    for k in range(n):
        if k == 0:
            fact = 1
        else:
            fact *= k
        total += 1 / fact
    return total

# 3 euler_approx
def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(total - true_e) > epsilon:
        if k == 0:
            fact = 1
        else:
            fact *= k
        total += 1 / fact
        k += 1
    return total

# 4 print_euler_sum_table
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(val)

# 测试
print(euler_infinite_sum(4))
```

```python
# ========== 判断整数幂 ==========
# 1 is_power_of_2 判断是不是2的幂
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

# 2 is_power(b, n) 判断 n 是否是 b 的幂
def is_power(b, n):
    if n == 1:
        return True
    while n % b == 0:
        n = n // b
    return n == 1

# 测试
print(is_power_of_2(8))
print(is_power_of_2(12))
print(is_power(3, 81))
print(is_power(3, 80))
```
